```python

# Data
days = ["Mon","Tue","Wed","Thur","Fri","Sat","Sun"]
mango_sales = [230,300,400,500,1000,985,650]
apple_sales = [340,123,234,598,934,980,1000]
guava_sales = [500,230,113,268,900,589,607]

plt.plot(days,mango_sales,marker="o",color="orange",linestyle="dashed",markersize=8,label="Mango")

plt.plot(days,apple_sales,marker="o",color="red",linestyle="dashed",markersize=8,label="Apple")

plt.plot(days,guava_sales,marker="o",color="green",linestyle="dashed",markersize=8,label="Guava")

for i in range(len(days)):
    plt.text(days[i],mango_sales[i] + 20,mango_sales[i],ha='center')
    plt.text(days[i],apple_sales[i] - 40,apple_sales[i],ha='center')
    plt.text(days[i],guava_sales[i] + 20,guava_sales[i],ha='center')

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Weekly Fruit Sales Comparison")

plt.gca().set_facecolor("lightblue")

plt.grid()

plt.legend(loc="best")

plt.ylim(0, 1200)
plt.show()
```

```python
#Task: Plot the weekly temperature of 2 cities — Mumbai and Delhi

days   = ["Mon", "Tue", "Wed", "Thur", "Fri", "Sat", "Sun"]
mumbai = [30, 32, 31, 33, 34, 35, 33]
delhi  = [22, 25, 27, 24, 28, 30, 26]

plt.plot(days,mumbai,marker ="o",color = "black",linestyle="dashed",markersize = 10,markerfacecolor="green",markeredgecolor = "yellow",label="Mumbai")
plt.plot(days,delhi,marker ="o",color = "black",linestyle="--",markersize = 10,markerfacecolor="red",markeredgecolor = "black",label="Delhi")

for i in range(len(days)):
    plt.text(days[i],mumbai[i] +1,mumbai[i],ha='center')
    plt.text(days[i],delhi[i] +1,delhi[i],ha='center')
    
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Comparison")

plt.gca().set_facecolor("grey")
plt.grid()
plt.legend(loc="best")
plt.ylim(15, 40)
plt.show()
```

```python
# Create a bar plot of 8 students and their marks in mathematics
# Customise it
# Also give some background colour , xlabel , ylabel , title

students = ["Aarav", "Zikra", "Rohit", "Priya", 
            "Sneha", "Karan", "Meera", "Arjun"]

marks = [85, 92, 67, 78, 95, 55, 88, 73]
bars = plt.bar(students,marks,color=["tomato","steelblue","gold","mediumseagreen",
                      "violet","coral","dodgerblue","orange"],
               edgecolor="black",)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Comparion of marks of each student in Maths")
plt.gca().set_facecolor("lightyellow")
plt.ylim(0, 110)

plt.bar_label(bars)
plt.grid(color = "black")
plt.show()

```

```python
students = ["Zikra","Bushra","Claire","Danish","Elle","Rory","Ganesh","Harry"]
marks = [96,45,33,67,89,25,56,15]

colors = ["green" if m > 50 else "red" for m in marks]

bars = plt.bar(students, marks, color=colors, edgecolor="black")

plt.xlabel("Students")
plt.ylabel("Marks in Mathematics")
plt.title("Math Marks of 8 Students")

plt.ylim(0, 120)
plt.bar_label(bars)

plt.gca().set_facecolor("lightyellow")
plt.grid(color = "black")
plt.show()
```
