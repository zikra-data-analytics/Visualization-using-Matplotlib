```python
# Create a scatter plot between mpg and hp (x and y)
# Make this plot with respect to the brand column
# use hue = "brand"
```

```python
sns.scatterplot(x='hp', y='mpg',hue='brand', data=df)
plt.title('MPG vs. Horsepower')
plt.xlabel('Hp')
plt.ylabel('Mpg')
plt.grid()
plt.show()
```

```python
#  Create line plot between hp-mpg , hp-time-to-60, hp -weightlbs
```

```python
sns.lineplot(data=df, x="hp", y="mpg", marker="o")
plt.title("Hp vs MPG")
plt.show()
```

```python
sns.lineplot(data=df, x="hp", y="time-to-60", marker="o", color="orange")
plt.title("Hp vs Time to 60 mph")
plt.show()
```

```python
sns.lineplot(data=df, x="hp", y="weightlbs", marker="o", color="green")
plt.title("Hp vs Weight (lbs)")
plt.show()
```

```python
# Create subplots

# MPG v HP
# Cylinders vs HP
# HP vs MPG
# HP vs Time-to-60

# scatterplot
```

```python
# Create subplots
fig = plt.subplots(2, 2, figsize=(12,10))

# step1: MPG vs HP
plt.subplot(2,2,1)
sns.scatterplot(data=df, x="mpg", y="hp", hue="brand")
plt.title("MPG vs HP")

# step2: Cylinders vs HP
plt.subplot(2,2,2)
sns.scatterplot(data=df, x="cylinders", y="hp", hue="brand")
plt.title("Cylinders vs HP")

# step3: HP vs MPG
plt.subplot(2,2,3)
sns.scatterplot(data=df, x="hp", y="mpg", hue="brand")
plt.title("HP vs MPG")

# step4: HP vs Time-to-60
plt.subplot(2,2,4)
sns.scatterplot(data=df, x="hp", y="time-to-60", hue="brand")
plt.title("HP vs Time-to-60")

plt.show()
```

```python
plt.subplot(2,2,1)
sns.scatterplot(
    data=df,
    x= "mpg",
    y="hp",
    hue = "brand",
    s=50,
    edgecolor="black",
    alpha=0.8,
    marker="o"
)
plt.title("MPG vs HP")

plt.subplot(2,2,2)
sns.scatterplot(
    data=df, 
    x= "cylinders",
    y="hp",
    hue = "brand",
    s=50,
    edgecolor="black",
    alpha=0.8,
    marker="o"
)
plt.title("Cylinders vs HP")

plt.subplot(2,2,3)
sns.scatterplot(
    data=df,
    x= "hp",
    y="mpg",
    hue = "brand",
    s=50,
    edgecolor="black",
    alpha=0.8,
    marker="o"
)
plt.title("HP Vs MPG")

plt.subplot(2,2,4)
sns.scatterplot(
    data=df,
    x="hp",
    y="time-to-60",
    hue="brand",
    s=50,
    edgecolor="black",
    alpha=0.8
)
plt.title("HP vs Time-to-60 ")

plt.grid(alpha=0.3)
plt.gca().set_facecolor("#eef2f7") 
plt.show()
```
